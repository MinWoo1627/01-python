# for 반복문
- 반복 횟수가 지정된 반복에 사용되는 구문
- python for문은 iterable(순회, 반복 가능) 객체의 요소를 하나씩 꺼내 실행하는 구문이다.
  - iterable 객체: list, tuple, set, dict, range 등...

# 1. 문자열, 리스트, 튜플 순회
- 시퀀스 타입 객체 (index가 존재한다)
- index 단위로 순회

````python
text: str = 'hello world'

for ch in text: # tet 문자열에서 한 글자씩 꺼내 ch에 대입
    print(ch)   # 모든 글자를 한 번씩 꺼내면 반복 종료

print('종료됨')
````

````python
# 리스트 순회
# 변수명: 타입힌트 = 값
lst: list[int] = [10, 20, 30, 40]

for num in lst:
    print(num)

print("종료됨")
````

````python
# 튜플 순회
tpl: tuple[int] = (100, 200, 300)
for num in tpl:
    print(num)
print("종료됨")
````

## 2. dict, set 순회

- dict를 `for...in` 구문을 이용해서 순회하면 key만 반환된다.
  - item 단위로 반환 받길 원하면 `dict.items()` 사용

- set은 순서가 없는 자료형이기 때문에 출력 순서는 실행 환경에 따라 달라진다.

````python
# dict 순회
from typing import Any # typing 패키지에서 Any 얻어오기

# Any : 어떤 타입이든 작성 가능하다는 것을 나타내는 힌트용 타입(클래스)
user_info: dict[str, Any] = {
    'name': '홍길동',
    'age': '32',
    'hobbies': ['운동','독서']
}

# key == 'name', 'age', 'hobbies'
for key in user_info:
    print(f"{key} : {user_info[key]}")

# [dict item 순회]
# for (key, value) in user_info.items() :
for key, value in user_info.items() :
    print(f"{key} : {value}")
````

````python
# set 순회
# - set 내부에 저장된 값을 꺼내는 유일한 방법
#   - set은 인덱스가 없기 때문에 원하는 값만 꺼내는 것이 불가능
nums: set[int] = {1, 2, 4, 6, 10, 15, 19}
for num in nums:
    print(num)
````

````python
# nums에 저장된 짝수만 찾아서 list에 옮겨담기
nums: set[int] = {1, 2, 4, 6, 10, 15, 19}

even_list: list[int] = [] # 짝수만 저장할 list
odd_list: list[int] = [] # 홀수만 저장할 list

for num in nums:
    if num % 2 ==0:
        even_list.append(num) # num 옮겨담기
    else:
        odd_list.append(num)

print(even_list)
print(odd_list)
````

## 3. range 객체
- 정수 범위를 표현하는 객체
- `range(start, end, step)`
  - start: 시작할 값
  - end: 끝 값 (미포함)
  - step: 증가 간격 (미작성 시 기본 값 1)

````python
rng = range(1, 5, 1) # 1 시작, 5 미만까지, 1씩 증가 -> (1,2,3,4)

print("rng: ", rng, type(rng))

for num in rng: # (1,2,3,4)
    print(num)
````

````python
# ragne를 이용한 0부터 10까지 추가되어있는 list 생성하기
nums: list[int] = list(range(0, 11))
print(nums)
````

````python
# 1부터 100까지의 정수 합 구하기
n_sum = 0 # 누적 수행 시 결과에 영향을 미치지 않게 하기 위해서 0 사용

for num in range(1, 101, 1):
    # n_sum = n_sum + num
    n_sum += num # 복합 대입 연산자

print(n_sum)
````

### (참고)  복합 대입 연산자
- 기존 변수를 이용한 연산 결과를 
  다시 기존 변수에 대입하는 연산자

- +=, -=, *=, /=, //=, %=, **= ... 등이 존재

```python
a = 10

# a에 20을 누적하는 방법

# 방법 1
a = a + 20

# 방법 2
a += 20

```


````python
a = 10
a += 5 # a= 10 + 5
print(a)

b = 100
b -= 30 # b = b - 30
print(b)

c = 4
c *= 5 # c = 4 * 5
print(c)

d = 7
d /= 2 # d = 7 / 2
print(d)
````

## 4. 중첩 반복문
- 반복문 안에 반복문이 존재하는 구조
- 다차원 구조를 생성하거나, 접근하는 용도로 많이 사용

시간 - 년월일시분초

````python
# 중첩 반복문 예시
for a in range(1,4):     # a = 1,2,3
    for b in range(1,3): # b - 1,2
        print(a,b)
````

````python
# 구구단 2단 출력하기
dan:int = 3 # 단 지정

for num in range(1, 10, 1):
    print(f"{dan} x {num} = {dan * num}")
````

````python
## 2단부터 9단까지 출력하기
dan:int = 3 # 단 지정
for dan in range(2, 10, 1):
    print(f"----- [{dan}] -----")
    for num in range(1, 10, 1):
        print(f"{dan} x {num} = {dan * num}")
````

````python
# 시작 단과 끝 단을 입력 받아서 해당 범위 만큼만 구구단 출력하기
""" 예시
시작 단 : 2
끝 단 : 4
----[2단]----
2 x 1 = 2
"""

start: int = int(input("시작 단: "))
end: int = int(input("끝 단: "))

for dan in range(start, end + 1, 1):
    print(f"----- [{dan}단] -----")
    for num in range(1, 10, 1):
        print(f"{dan} x {num} = {dan * num}")
````
