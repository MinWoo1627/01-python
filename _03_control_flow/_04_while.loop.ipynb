# while 반복문

- 조건식이 True인 동안에만 반복하는 반복문
  - False가 되는 경우 반복 종료

- 반복 횟수가 정해지지 않는 반복에 사용하는 것이 적절

## 1. 기본 while 구조
```python
while 조건식:
    # 조건식이 True 일 때 반복할 구문

```

````python
# 1부터 10까지 1씩 증가시키면서 (10번) i값을 출력하는 기본 반복문
i = 1

while i <= 10:
    print(i)
    i += 1
````

````python
# 10 ~ 1 역순 출력
i = 10

while i > 0: # i가 0보다 큰 경우 True 였다가 0이 되는 순간 False
    print(i)
    i -= 1 
````

## 2. 분기문

- 반복문 실행 중 흐름을 제어하는 구문
- `break` : break가 포함된 반복문을 즉시 반복 종료
- `continue` : continue가 포함된 반복문의 다음 회차로 이동

````python
# break
# 6 까지 출력 후 멈추기

i = 1

while i <= 10: 
    print(i)
    if i == 6:
        break
    i += 1
````

````python
# 무한 루프 멈추기

# 입력 받은 양의 정수가 되면 멈추기

destination = int(input('양의 정수를 입력해:'))


i = 1

while True: # 무한 루프
    print(i)
    if(i == destination):
        break
    i += 1
````

````python
# continue
# 1 ~ 20까지 순서대로 출력
# 단, 홀수만 출력

for num in range(1, 21, 1):

    if num % 2 == 0:
        continue

    print(num, end=" ") # end -> print 종료 후 마지막 문자 지정
                        # 기본값: 엔터(\n)
````

````python
# 반복문 + 조건문 + 분기문
# 예시 1)

orders: list[str] = ['아메리카노', '취소', '라떼', '종료', '녹차']

for order in orders:
    if order == '취소':
        continue
    if order == '종료':
        break
    print(f"{order}가 준비되었습니다!")
print("종료되었습니다.")
````

````python
# 반복문 + 조건문 + 분기문
# 예시 2)

order_list = []

while True:

    # str.strip() 문자열 앞 뒤의 공백을 없앤다.

    order = input('메뉴를 입력하세요 (종료: exit 입력): ').strip()
    
    if order == '':
        continue

    # str.lower() : 소문자로 변환
    if order.lower() == 'exit': # 반복 입력 종료
        print("<<입력 종료>>")
        break
    order_list.append(order)

print(order_list)
````
